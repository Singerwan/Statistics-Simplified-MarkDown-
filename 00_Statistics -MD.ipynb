<span style="font-size:25px;background-color:red">Normalization</span>
<hr>
zscore normalziation -why
<hr>
<ol>
    <li>units are  different | height vs weight | transform <span style="background-color:red"> unit less </li>
    <li>formula</li>
        <ul>
            <li style="font-size:25px;color:cyan">  $z_{i}=\frac{x_{i}-\bar x }{\sigma_{x}}$ </li>
            <li>how many standard deviation is the distance away from the center mean </li>
            <li>center - dispersion zscore transformation needs to be a normal distribution</li>
        </ul>
</ol>
<div style="font-size:25px;background-color:red">             
 $\bar x =\frac{x-x_{mim}}{x_{max}-x_{mim}}$ &emsp;$\bar x =\frac{x-x_{mim}}{range}$
    <span style="font-size:15px">scale to a range of 0 to 1</div>
